```python
# List comprehension
result = [x ** 2 for x in range(10) if x % 2 == 0]
print(result)

# Dictionary comprehension
dict_comp = {x: x ** 2 for x in range(5)}
print(dict_comp)

# Set comprehension
set_comp = {x % 3 for x in range(10)}
print(set_comp)
```

```python
# Generators
def count_up_to(n):
    i = 0
    while i < n:
        yield i
        i += 1

for num in count_up_to(5):
    print(num)
```
