# Python Internals

## What You'll Learn
- Python object model
- Memory management
- Garbage collection
- CPython internals

```python
# Understanding Python objects
import sys

# Every object has reference count
obj = []
print(f"Reference count: {sys.getrefcount(obj)}")

# Object IDs
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(f"id(a) = {id(a)}")
print(f"id(b) = {id(b)}")
print(f"id(c) = {id(c)}")
print(f"a is b: {a is b}")
print(f"a is c: {a is c}")
```

```python
# Garbage collection
import gc

print(f"Garbage collection enabled: {gc.isenabled()}")
print(f"Collection counts: {gc.get_count()}")

# Force garbage collection
gc.collect()
print("Garbage collected")
```
