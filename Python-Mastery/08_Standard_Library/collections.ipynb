```python
# Collections
from collections import Counter, defaultdict, namedtuple

# Counter
c = Counter(['a', 'b', 'a', 'c', 'b', 'a'])
print(c)  # Counter({'a': 3, 'b': 2, 'c': 1})
print(c.most_common(2))  # [('a', 3), ('b', 2)]
```

```python
# defaultdict
from collections import defaultdict

d = defaultdict(int)
d['a'] += 1
d['b'] += 2
print(dict(d))  # {'a': 1, 'b': 2}
```
