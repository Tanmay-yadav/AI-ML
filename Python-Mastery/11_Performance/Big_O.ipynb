```python
# Timing with timeit
import timeit

# Compare two approaches
setup = "x = list(range(100))"

time1 = timeit.timeit('max(x)', setup=setup, number=100000)
print(f"max() function: {time1:.4f}s")

time2 = timeit.timeit(
    'sorted(x, reverse=True)[0]', 
    setup=setup, 
    number=100000
)
print(f"sorted() approach: {time2:.4f}s")
```

```python
# Memoization with lru_cache
from functools import lru_cache

@lru_cache(maxsize=128)
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(30))
print(f"Cache info: {fibonacci.cache_info()}")
```
