```python
# Writing to a file
with open('example.txt', 'w') as f:
    f.write('Hello, World!\n')
    f.write('Line 2\n')

print("File written successfully")
```

```python
# Reading from a file
with open('example.txt', 'r') as f:
    content = f.read()
    print(content)

# Read line by line
with open('example.txt', 'r') as f:
    for line in f:
        print(line.strip())
```

```python
# JSON files
import json

data = {'name': 'Alice', 'age': 25}

# Write JSON
with open('data.json', 'w') as f:
    json.dump(data, f)

# Read JSON
with open('data.json', 'r') as f:
    loaded_data = json.load(f)
    print(loaded_data)
```
