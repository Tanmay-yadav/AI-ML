# Packaging and Distribution

## What You'll Learn
- Creating Python packages
- setup.py and pyproject.toml
- Publishing to PyPI
- Version management

## Project Structure

```
my_package/
├── my_package/
│   ├── __init__.py
│   ├── module1.py
│   └── module2.py
├── tests/
│   ├── __init__.py
│   └── test_module1.py
├── README.md
├── setup.py
├── pyproject.toml
└── requirements.txt
```

## setup.py Example

```python
from setuptools import setup, find_packages

setup(
    name="my-package",
    version="0.1.0",
    description="A short description",
    author="Your Name",
    author_email="email@example.com",
    packages=find_packages(),
    install_requires=[
        "requests>=2.20.0",
    ],
    python_requires=">=3.6",
)
```

## Publishing to PyPI

1. Create account on PyPI
2. Build distribution: `python setup.py sdist bdist_wheel`
3. Upload: `twine upload dist/*`
