# Type Conversion

## What You'll Learn
- Converting between data types
- Implicit vs explicit conversion
- Common conversions

```python
# Explicit type conversion
string_num = "42"
integer = int(string_num)
print(f"String: {string_num} (type: {type(string_num)})")
print(f"Integer: {integer} (type: {type(integer)})")

# Float conversion
float_num = float("3.14")
print(f"Float: {float_num}")

# String conversion
num = 42
string = str(num)
print(f"String: {string} (type: {type(string)})")
```
