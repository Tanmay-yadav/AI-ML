```python
# Numeric types
integer = 42
floating = 3.14
complex_num = 3 + 4j

# String type
string = "Hello, Python!"

# Boolean type
boolean = True

# Check types
print(f"Type of integer: {type(integer)}")
print(f"Type of floating: {type(floating)}")
print(f"Type of string: {type(string)}")
print(f"Type of boolean: {type(boolean)}")
```

## Collection Data Types

| Type | Syntax | Example | Mutable |
|------|--------|---------|----------|
| List | `[]` | `[1, 2, 3]` | Yes |
| Tuple | `()` | `(1, 2, 3)` | No |
| Set | `{}` | `{1, 2, 3}` | Yes |
| Dictionary | `{}` | `{'a': 1}` | Yes |
| String | `""` | `"hello"` | No |
