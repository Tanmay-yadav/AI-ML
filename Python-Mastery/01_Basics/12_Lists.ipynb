```python
# Creating lists
empty_list = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]
nested = [[1, 2], [3, 4], [5, 6]]

print(numbers)
print(nested)
```

```python
# List operations
numbers = [1, 2, 3, 4, 5]
print(f"Length: {len(numbers)}")
print(f"First: {numbers[0]}")
print(f"Last: {numbers[-1]}")
print(f"Slice [1:3]: {numbers[1:3]}")
```

```python
# List methods
numbers = [1, 2, 3]
numbers.append(4)          # Add to end
numbers.extend([5, 6])     # Add multiple
numbers.insert(0, 0)       # Insert at index
print(f"After operations: {numbers}")

numbers.remove(3)          # Remove by value
print(f"After remove: {numbers}")
print(f"Index of 2: {numbers.index(2)}")
```
