# Sets

## What You'll Learn
- Creating sets
- Set operations
- Set methods
- Union, intersection, difference

```python
# Creating sets
empty_set = set()          # {} is for empty dict
numbers = {1, 2, 3, 4, 5}
mixed = {1, "hello", 3.14}

# Sets don't have duplicates
duplicates = {1, 1, 2, 2, 3}
print(f"Unique values: {duplicates}")
```

```python
# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(f"Union: {a | b}")
print(f"Intersection: {a & b}")
print(f"Difference: {a - b}")
print(f"Symmetric difference: {a ^ b}")
```

```python
# Set methods
s = {1, 2, 3}
s.add(4)
print(f"After add: {s}")

s.remove(2)         # Raises error if not found
print(f"After remove: {s}")

s.discard(5)        # No error if not found
print(f"After discard: {s}")
```
