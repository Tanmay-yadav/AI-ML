# None

## What You'll Learn
- What is None
- Using None
- None vs False vs 0

```python
# None is Python's null value
x = None

print(f"Value: {x}")
print(f"Type: {type(x)}")
print(f"Is None: {x is None}")
print(f"Is not None: {x is not None}")
```

```python
# None vs other falsy values
print(None == False)      # False (not equal)
print(None == 0)          # False
print(None is False)      # False (not identical)
print(bool(None))         # False (None is falsy)
```
