# Booleans

## What You'll Learn
- Boolean values
- Truthy and falsy values
- Boolean operations

```python
# Boolean values
is_true = True
is_false = False

print(f"True: {is_true}")
print(f"False: {is_false}")
print(f"Type: {type(is_true)}")
```

```python
# Truthy and falsy values
# Falsy values: False, 0, 0.0, '', [], {}, None
# Truthy values: True, any non-zero number, non-empty string/list/dict, etc.

print(bool(0))        # False
print(bool(1))        # True
print(bool(""))       # False
print(bool("hello"))  # True
print(bool([]))       # False
print(bool([1, 2]))   # True
```
