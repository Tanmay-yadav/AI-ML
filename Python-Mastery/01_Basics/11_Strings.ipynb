# Strings

## What You'll Learn
- Creating strings
- String operations
- String methods
- Escape sequences

```python
# Creating strings
single_quote = 'Hello'
double_quote = "Hello"
multiline = """This is
a multiline
string"""

print(single_quote)
print(double_quote)
print(multiline)
```

```python
# String indexing and slicing
text = "Python"
print(f"First character: {text[0]}")
print(f"Last character: {text[-1]}")
print(f"Slice [1:4]: {text[1:4]}")
print(f"Every other character: {text[::2]}")
```

```python
# String methods
text = "hello world"
print(f"Upper: {text.upper()}")
print(f"Capitalize: {text.capitalize()}")
print(f"Replace: {text.replace('world', 'python')}")
print(f"Split: {text.split()}")
print(f"Find: {text.find('world')}")
```
