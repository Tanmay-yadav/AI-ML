```python
# Creating tuples
empty_tuple = ()
single_item = (1,)           # Comma makes it a tuple
numbers = (1, 2, 3, 4, 5)
mixed = (1, "hello", 3.14)

print(numbers)
print(type(single_item))
```

```python
# Tuples are immutable (cannot be changed)
t = (1, 2, 3)
# t[0] = 10  # This would raise an error
# t.append(4)  # This would raise an error

print(f"Length: {len(t)}")
print(f"First: {t[0]}")
print(f"Count of 2: {t.count(2)}")
```
