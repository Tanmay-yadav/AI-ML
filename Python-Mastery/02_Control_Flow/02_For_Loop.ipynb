# For Loops

## What You'll Learn
- Iterating with for loops
- Range function
- Looping through sequences
- Enumerate and zip

```python
# Simple for loop
for i in range(5):
    print(f"Number: {i}")
```

```python
# Looping through sequences
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"Fruit: {fruit}")
```

```python
# Enumerate - get index and value
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

```python
# Zip - combine sequences
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(f"{name}: {age}")
```
