```python
# Basic while loop
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1
```

```python
# While with complex condition
password_attempts = 0
max_attempts = 3
correct_password = "secret123"

while password_attempts < max_attempts:
    # password = input("Enter password: ")
    password = "secret123"  # For demo
    if password == correct_password:
        print("Access granted!")
        break
    else:
        password_attempts += 1
        print(f"Wrong password. Attempts remaining: {max_attempts - password_attempts}")
```
