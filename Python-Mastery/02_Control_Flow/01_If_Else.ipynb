# If Else Statements

## What You'll Learn
- Conditional statements
- If, elif, else
- Nested conditions
- Ternary operator

```python
# Basic if-else
age = 25

if age >= 18:
    print("You are an adult")
else:
    print("You are a minor")
```

```python
# If-elif-else
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Grade: {grade}")
```

```python
# Ternary operator
age = 25
status = "Adult" if age >= 18 else "Minor"
print(status)
```
