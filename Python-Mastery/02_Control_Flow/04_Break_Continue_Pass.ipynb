# Break, Continue, Pass

## What You'll Learn
- Break statement
- Continue statement
- Pass statement

```python
# Break - exit loop
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# Continue - skip to next iteration
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)  # Only odd numbers
```

```python
# Pass - placeholder (do nothing)
for i in range(5):
    if i == 2:
        pass  # Placeholder for future code
    else:
        print(i)
```
