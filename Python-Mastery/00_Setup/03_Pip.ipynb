# Pip - Python Package Manager

## What You'll Learn
- Installing packages with pip
- Managing package versions
- Working with requirements.txt
- Common pip commands

## Common pip Commands

### Installing Packages
```bash
pip install package_name
pip install package_name==1.2.3  # Specific version
pip install package_name>=1.0    # Minimum version
```

### Managing Packages
```bash
pip list                         # List installed packages
pip show package_name            # Show package info
pip uninstall package_name       # Remove package
pip install --upgrade package    # Update package
```
