## Installing and Running Jupyter

### Installation
```bash
pip install jupyter
pip install jupyterlab  # More advanced
```

### Running Jupyter
```bash
jupyter notebook       # Classic Notebook
jupyter lab           # JupyterLab
```

## Keyboard Shortcuts

### Command Mode (press Esc)
- `y` - Convert to code cell
- `m` - Convert to markdown cell
- `a` - Insert cell above
- `b` - Insert cell below
- `d,d` - Delete cell
- `z` - Undo delete
