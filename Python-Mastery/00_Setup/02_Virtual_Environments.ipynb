## What is a Virtual Environment?

A virtual environment is an isolated Python workspace that allows you to:
- Install packages without affecting system Python
- Maintain different package versions for different projects
- Avoid dependency conflicts
- Make your project portable and reproducible

## Creating a Virtual Environment

### Using venv (built-in)
```bash
python -m venv myenv
source myenv/bin/activate  # On Linux/macOS
myenv\Scripts\activate     # On Windows
```

### Using conda
```bash
conda create -n myenv python=3.11
conda activate myenv
```
