## Project Template

Each project folder should contain:
- `README.md` - Project description
- `requirements.txt` - Dependencies
- `main.py` or `app.py` - Entry point
- `src/` - Source code modules
- `tests/` - Test files
- `.gitignore` - Files to ignore in git
