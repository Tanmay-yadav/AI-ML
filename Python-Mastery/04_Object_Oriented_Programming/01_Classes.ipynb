## Classes - The Foundation

A class is a blueprint for creating objects.

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
    
    def bark(self):
        return f"{self.name} says Woof!"

# Create an instance
dog = Dog("Buddy", "Golden Retriever")
print(dog.bark())
```
