# Args and Kwargs

## What You'll Learn
- *args for variable positional arguments
- **kwargs for variable keyword arguments
- Combining with regular parameters

```python
# *args - accept variable number of positional arguments
def sum_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

print(sum_numbers(1, 2, 3))
print(sum_numbers(1, 2, 3, 4, 5))
print(f"Args type: {type((1, 2, 3))}")
```

```python
# **kwargs - accept variable number of keyword arguments
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=25, city="NYC")
```

```python
# Combining regular parameters, *args, and **kwargs
def print_everything(name, *args, **kwargs):
    print(f"Name: {name}")
    print(f"Args: {args}")
    print(f"Kwargs: {kwargs}")

print_everything("Alice", 1, 2, 3, city="NYC", age=25)
```
