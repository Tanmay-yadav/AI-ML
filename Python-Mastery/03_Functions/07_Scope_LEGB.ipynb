# Scope - LEGB Rule

## What You'll Learn
- Local scope
- Enclosing scope
- Global scope
- Built-in scope
- LEGB rule explanation

```python
# LEGB = Local, Enclosing, Global, Built-in

# Global scope
x = "Global"

def outer():
    # Enclosing scope
    x = "Enclosing"
    
    def inner():
        # Local scope
        x = "Local"
        print(x)  # Local
    
    inner()
    print(x)  # Enclosing

outer()
print(x)  # Global
```

```python
# Global and nonlocal keywords
x = 10

def modify_global():
    global x
    x = 20

print(f"Before: {x}")
modify_global()
print(f"After: {x}")
```
