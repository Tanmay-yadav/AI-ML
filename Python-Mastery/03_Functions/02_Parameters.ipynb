# Parameters

## What You'll Learn
- Positional parameters
- Keyword parameters
- Default values
- Variable-length parameters (*args, **kwargs)

```python
# Positional vs Keyword arguments
def person_info(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

# Positional
person_info("Alice", 25, "NYC")

# Keyword
person_info(name="Bob", city="LA", age=30)

# Mixed
person_info("Charlie", age=35, city="SF")
```

```python
# Default values
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", "Hi"))
```
