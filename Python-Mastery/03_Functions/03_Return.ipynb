# Return Statement

## What You'll Learn
- Returning values
- Multiple return values
- Early returns

```python
# Simple return
def square(x):
    return x ** 2

result = square(5)
print(f"5^2 = {result}")
```

```python
# Multiple return values
def get_coordinates():
    x = 10
    y = 20
    return x, y  # Returns a tuple

x_coord, y_coord = get_coordinates()
print(f"Coordinates: ({x_coord}, {y_coord})")
```

```python
# Early return
def validate_age(age):
    if age < 0:
        return "Age cannot be negative"
    if age > 150:
        return "Age seems unrealistic"
    return "Age is valid"

print(validate_age(-5))
print(validate_age(25))
print(validate_age(200))
```
