# Curse Of Dimensionality 
  - The Higher the Dimensionality, the higher the sparsity which leads to the decrease in performance.
  - And it increases the Computaion.
  - The solving method is by Dimensionality reduction :
    - Feature Selection
      - Foreward Selection
      - Backward selection
    - Feature Extraction 
      - PCA
      - LDA
      - TSNE
  - Higher Dimesional Data is Dangerous for ML Algorithms.
