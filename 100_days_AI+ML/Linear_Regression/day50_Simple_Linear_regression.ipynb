# Linear Regression
- Linear Regression is a supervised machine learning  algorithm.
  - `Types of Linear Regression`
    - Simple Linear Regression (Single input column)
    - Multiple Linear Regression(Multiple Input columns)
    - Polynomial Linear Regression (When Data is not linear)

    ## Simple Linear Regression
    A Simple example of simple linear regression is predicting package on the basis of cgpa.
    In simple linear regression we have to find the best fit line which has the least error.
    The Line would be : `y = mx +b` 
    - `m`: is the slope or weight of the input variable x
    - `x`: is the input column
    - `b`: is the bias 


```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('E:\\WORK\\AI-ML\\data\\placement.csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'], df['package'])
plt.xlabel('CGPA')
plt.ylabel('package')
plt.title('Relationship between CGPA and Salary')
plt.show()
```

```python
x =df.iloc[:,0:1]
y =df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python

lr = LinearRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
x_test
```

```python
y_test
```

```python
lr.predict(x_test.iloc[0].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_train,lr.predict(x_train),color='red')
plt.xlabel('CGPA')
plt.ylabel('Package(in lpa)')
```

```python
m = lr.coef_
```

```python
b = lr.intercept_
```

```python
m * 9.5 + b
```

```python
m * 100 + b
```
