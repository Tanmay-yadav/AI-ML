# what are Outliers? 
- Outliers are the values that behave very differently from all the other data points.
## Effects of Ouliers on ML algorithms?
- it can change the mean value and leads to wrong predictions by the Algorithms.
- In algorithms like linear regression , Adaboost,Logistic Regression and Deep Learning  outliers plays a major impact.

## How to treat Outliers?
- Trimming: remove the outliers.
- Capping : Limiting the data
- Treat the Outliers as missing values
- Discritization: Dividing the data into fixed intervals.like data can be divided into the brackets of 10 and if 90 is an outlier then it comes between a bracket of 90-100.
