# Eigenvalues, Eigenvectors, Linear Transformations, and PCA

## Q1. What is a Linear Transformation?

A linear transformation is a function that transforms vectors while preserving vector addition and scalar multiplication.

A transformation $T$ is linear if:

$$T(\mathbf{u}+\mathbf{v})=T(\mathbf{u})+T(\mathbf{v})$$

and

$$T(c\mathbf{u})=cT(\mathbf{u})$$

A matrix can represent a linear transformation.

For example:

$$A=\begin{bmatrix}2&0\\0&2\end{bmatrix}$$

transforms:

$$\begin{bmatrix}x\\y\end{bmatrix}\rightarrow\begin{bmatrix}2x\\2y\end{bmatrix}$$

This simply scales the vector by 2.

---

## Q2. What are Eigenvalues and Eigenvectors?

For a matrix $A$, an eigenvector is a non-zero vector whose direction does not change when the matrix is applied to it.

The mathematical definition is:

$$A\mathbf{v}=\lambda\mathbf{v}$$

where:

- $A$ = transformation matrix
- $\mathbf{v}$ = eigenvector
- $\lambda$ = eigenvalue

The eigenvalue tells us how much the eigenvector is scaled.

Therefore:

$$\boxed{\text{Eigenvector = special direction}}$$

$$\boxed{\text{Eigenvalue = scaling factor}}$$

---

## Q3. Where does the matrix $A$ come from?

In general linear algebra, $A$ is a matrix that represents a particular linear transformation.

You can be given or choose a matrix such as:

$$A=\begin{bmatrix}2&0\\0&3\end{bmatrix}$$

This matrix defines how vectors will be transformed.

In PCA specifically, $A$ is usually the covariance matrix of the dataset.

So in PCA:

$$\boxed{A=\text{covariance matrix}}$$

---

## Q4. Where does the vector $v$ come from?

In general, $v$ can be any non-zero vector that we want to test against the transformation.

For example:

$$v=\begin{bmatrix}1\\2\end{bmatrix}$$

We apply the matrix:

$$Av$$

In PCA, however, we are specifically looking for vectors $v$ that satisfy:

$$Av=\lambda v$$

These special vectors are the eigenvectors of the covariance matrix.

So in PCA:

$$\boxed{v=\text{an eigenvector of the covariance matrix}}$$

and it represents a direction in the feature space.

---

## Q5. How can the direction of a vector change after applying a matrix?

Suppose:

$$A=\begin{bmatrix}2&0\\0&3\end{bmatrix}$$

and:

$$v=\begin{bmatrix}1\\1\end{bmatrix}$$

Then:

$$Av=\begin{bmatrix}2&0\\0&3\end{bmatrix}\begin{bmatrix}1\\1\end{bmatrix}=\begin{bmatrix}2\\3\end{bmatrix}$$

Originally the vector was:

$$(1,1)$$

After transformation it became:

$$(2,3)$$

The direction changed because:

$$\frac{1}{1}=1$$

but:

$$\frac{3}{2}=1.5$$

Therefore, $v=[1,1]$ is not an eigenvector of $A$.

---

## Q6. Are vectors whose direction changes after transformation eigenvectors?

No.

If applying $A$ changes the direction of $v$, then $v$ is not an eigenvector.

An eigenvector must satisfy:

$$Av=\lambda v$$

For example:

$$A=\begin{bmatrix}2&0\\0&3\end{bmatrix}$$

and:

$$v=\begin{bmatrix}1\\0\end{bmatrix}$$

Then:

$$Av=\begin{bmatrix}2\\0\end{bmatrix}=2\begin{bmatrix}1\\0\end{bmatrix}$$

Therefore $v$ is an eigenvector and:

$$\lambda=2$$

---

## Q7. Does every vector have an eigenvalue?

No.

Eigenvalues are associated with eigenvectors of a matrix.

For a vector $v$ to have a corresponding eigenvalue $\lambda$, it must satisfy:

$$Av=\lambda v$$

For example:

$$A=\begin{bmatrix}2&0\\0&3\end{bmatrix}$$

and:

$$v=\begin{bmatrix}1\\1\end{bmatrix}$$

give:

$$Av=\begin{bmatrix}2\\3\end{bmatrix}$$

There is no single $\lambda$ such that:

$$\begin{bmatrix}2\\3\end{bmatrix}=\lambda\begin{bmatrix}1\\1\end{bmatrix}$$

Therefore, this vector is not an eigenvector and has no corresponding eigenvalue.

---

## Q8. What is the determinant?

The determinant is a single number calculated from a square matrix.

For a $2\times2$ matrix:

$$A=\begin{bmatrix}a&b\\c&d\end{bmatrix}$$

the determinant is:

$$\boxed{\det(A)=ad-bc}$$

For example:

$$A=\begin{bmatrix}2&0\\0&3\end{bmatrix}$$

Then:

$$\det(A)=(2)(3)-(0)(0)=6$$

---

## Q9. What is $I$ in $\det(A-\lambda I)=0$?

$I$ is the identity matrix.

For a $2\times2$ matrix:

$$I=\begin{bmatrix}1&0\\0&1\end{bmatrix}$$

Therefore:

$$\lambda I=\begin{bmatrix}\lambda&0\\0&\lambda\end{bmatrix}$$

So:

$$A-\lambda I$$

means subtracting $\lambda$ from the diagonal elements of $A$.

---

## Q10. Why do we use $\det(A-\lambda I)=0$ to find eigenvalues?

Start with the definition of an eigenvector:

$$Av=\lambda v$$

Move everything to one side:

$$Av-\lambda v=0$$

Since:

$$v=Iv$$

we can write:

$$Av-\lambda Iv=0$$

Factor out $v$:

$$(A-\lambda I)v=0$$

We want a non-zero vector $v$ to satisfy this equation.

For that to be possible, $A-\lambda I$ must be singular.

A matrix is singular when its determinant is zero.

Therefore:

$$\boxed{\det(A-\lambda I)=0}$$

This equation allows us to find the possible eigenvalues.

---

## Q11. What is $A$, $v$, and $\lambda$ in Machine Learning?

In PCA:

$$\boxed{A=\text{covariance matrix}}$$

$$\boxed{v=\text{eigenvector / principal direction}}$$

$$\boxed{\lambda=\text{eigenvalue / variance along that direction}}$$

For example, if your features are:

- Height
- Weight

then the covariance matrix could be:

$$A=\begin{bmatrix}\operatorname{Var}(\text{Height})&\operatorname{Cov}(\text{Height},\text{Weight})\\\operatorname{Cov}(\text{Height},\text{Weight})&\operatorname{Var}(\text{Weight})\end{bmatrix}$$

The eigenvectors of this matrix represent directions in the Height-Weight feature space.

The eigenvalues tell us how much variance exists in those directions.

---

## Q12. How is the eigenvector $v$ calculated for a Height-Weight dataset?

Suppose the centered dataset is:

| Person | Height | Weight |
|---|---:|---:|
| 1 | -5 | -5 |
| 2 | 0 | 0 |
| 3 | 5 | 5 |

Its covariance matrix is:

$$A=\begin{bmatrix}25&25\\25&25\end{bmatrix}$$

First, find the eigenvalues using:

$$\det(A-\lambda I)=0$$

This gives:

$$\lambda_1=50$$

and:

$$\lambda_2=0$$

Then, for each eigenvalue, solve:

$$(A-\lambda I)v=0$$

For:

$$\lambda_1=50$$

we get:

$$\begin{bmatrix}-25&25\\25&-25\end{bmatrix}\begin{bmatrix}v_1\\v_2\end{bmatrix}=\begin{bmatrix}0\\0\end{bmatrix}$$

This gives:

$$v_1=v_2$$

So one possible eigenvector is:

$$v=\begin{bmatrix}1\\1\end{bmatrix}$$

This represents a direction where Height and Weight increase together.

---

## Q13. Why does $[1,1]$ become $[0.707,0.707]$?

Because $[1,1]$ is normalized.

The length of:

$$v=\begin{bmatrix}1\\1\end{bmatrix}$$

is:

$$\|v\|=\sqrt{1^2+1^2}=\sqrt2$$

To make the vector have length 1, divide each element by $\sqrt2$:

$$\frac{1}{\sqrt2}\begin{bmatrix}1\\1\end{bmatrix}$$

Since:

$$\frac{1}{\sqrt2}\approx0.707$$

we get:

$$\boxed{v=\begin{bmatrix}0.707\\0.707\end{bmatrix}}$$

Both vectors:

$$\begin{bmatrix}1\\1\end{bmatrix}$$

and:

$$\begin{bmatrix}0.707\\0.707\end{bmatrix}$$

point in exactly the same direction.

They differ only in length.

PCA usually uses the normalized version, whose length is 1.

---

## Q14. What do eigenvectors and eigenvalues mean in PCA?

Eigenvectors represent the directions in which the data varies.

Eigenvalues represent how much variance the data has along those directions.

For example:

$$v_1=\begin{bmatrix}0.707\\0.707\end{bmatrix}$$

with:

$$\lambda_1=50$$

means that the data has a large amount of variation in the direction where Height and Weight increase together.

PCA sorts the eigenvectors by their eigenvalues.

The eigenvector with the largest eigenvalue becomes the first principal component direction.

Therefore:

$$\boxed{\text{Largest eigenvalue}\rightarrow\text{Most variance}\rightarrow\text{First principal component}}$$

---

## Q15. Why does PCA produce uncorrelated principal components?

PCA uses the eigenvectors of the covariance matrix.

The eigenvectors corresponding to different eigenvalues are orthogonal.

When the data is projected onto these orthogonal directions, the resulting principal components have zero covariance with each other.

Therefore:

$$\operatorname{Cov}(PC_1,PC_2)=0$$

So PCA produces principal components that are uncorrelated.

Important: uncorrelated does not necessarily mean statistically independent. It means they have zero linear correlation.



# Variance, Covariance, Correlation, and Covariance Matrix

## Q1. What is covariance? Does covariance mean the relationship between two features of a dataset?

### Answer

**Covariance measures how two variables/features vary together.**

It answers:

 When one variable moves away from its mean, does the other variable tend to move in the same direction or the opposite direction?

There are three cases:

| Covariance | Meaning |
|---|---|
| Positive | Both variables tend to move in the same direction |
| Negative | Variables tend to move in opposite directions |
| Near 0 | Little or no linear co-movement |

For example:

- Hours studied ↑ and exam score ↑ → positive covariance
- Price ↑ and quantity demanded ↓ → negative covariance

The formula for sample covariance is:

\[
Cov(X,Y)=\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{n-1}
\]

---

## Q2. What is variance?

### Answer

**Variance measures how spread out the values of one variable are around its mean.**

For example:

\[
X=[2,4,6]
\]

Mean:

\[
\bar X=4
\]

Deviations:

\[
[-2,0,+2]
\]

Squared deviations:

\[
[4,0,4]
\]

Variance measures the average of these squared deviations.

For a sample:

\[
Var(X)=\frac{\sum(X_i-\bar X)^2}{n-1}
\]

So:

 **Variance measures the spread of one variable, while covariance measures how two variables vary together.**

---

## Q3. Does variance only tell us the change in one variable or one axis?

### Answer

Yes, conceptually.

Variance deals with **one variable at a time**.

If \(X\) is the horizontal axis:

```text
X axis
2 ---- 4 ---- 6
       ↑
      mean
```

Variance asks:

- How spread out are the X values around their mean?

It does not require another variable.

---

## Q4. What is the major difference between variance and covariance?

### Answer

| | Variance | Covariance |
|---|---|---|
| Number of variables | 1 | 2 |
| Measures | Spread of one variable | Joint variation of two variables |
| Example | Spread of Age | How Age and Salary vary together |
| Can it be negative? | No | Yes |
| Formula | \(\frac{\sum(X_i-\bar X)^2}{n-1}\) | \(\frac{\sum(X_i-\bar X)(Y_i-\bar Y)}{n-1}\) |

A useful relationship is:

\[
\boxed{Var(X)=Cov(X,X)}
\]

Variance is therefore a special case of covariance where we compare a variable with itself.

---

# Correlation

## Q5. Why can't we say that covariance tells us exactly how strongly two features are related?

### Answer

Covariance does tell us about **joint variation**, but its numerical magnitude depends on the **scale and units** of the variables.

For example:

\[
X=[1,2,3]
\]

and:

\[
Y=[100,200,300]
\]

There is a perfect linear relationship:

\[
Y=100X
\]

Now multiply \(X\) by 100:

\[
X'=[100,200,300]
\]

The underlying relationship has not changed. We only changed the scale.

However, the covariance changes because covariance depends on the magnitude of the deviations.

Therefore:

> A larger covariance value does not necessarily mean a stronger relationship.

For example, changing meters to centimeters can make covariance numerically much larger even though the actual relationship between the variables has not changed.

---

## Q6. What tells us the strength of the relationship between two variables?

### Answer

**Correlation**, especially Pearson correlation for linear relationships.

The formula is:

\[
r=\frac{Cov(X,Y)}{\sigma_X\sigma_Y}
\]

where:

- \(Cov(X,Y)\) = covariance
- \(\sigma_X\) = standard deviation of X
- \(\sigma_Y\) = standard deviation of Y

Correlation is standardized and always lies between:

\[
-1\leq r\leq1
\]

Typical interpretation:

| Correlation | Meaning |
|---:|---|
| +1 | Perfect positive linear relationship |
| +0.8 | Strong positive linear relationship |
| +0.3 | Weak/moderate positive linear relationship |
| 0 | No linear relationship |
| -0.8 | Strong negative linear relationship |
| -1 | Perfect negative linear relationship |

Important:

> Correlation measures linear association. It does not prove causation.

---

# Scaling and Linear Transformation

## Q7. How can covariance change if the vector changes only by scaling?

### Answer

Consider:

\[
X=[1,2,3]
\]

and:

\[
Y=[100,200,300]
\]

Their relationship is:

\[
Y=100X
\]

Now scale \(X\) by 100:

\[
X'=100X=[100,200,300]
\]

The relationship is still perfectly linear.

However, the deviations of \(X\) from its mean become 100 times larger.

Original:

\[
X-\bar X=[-1,0,+1]
\]

After scaling:

\[
X'-\bar X'=[-100,0,+100]
\]

Because covariance multiplies deviations:

\[
Cov(X,Y)
=
\frac{\sum(X_i-\bar X)(Y_i-\bar Y)}{n-1}
\]

the covariance becomes larger.

In this example:

\[
Cov(X,Y)=100
\]

but:

\[
Cov(X',Y)=10000
\]

The relationship did not become 100 times stronger.

Only the **scale of X changed**.

---

## Q8. What is a linear transformation?

### Answer

A simple scaling transformation is:

\[
X'=aX
\]

For example:

\[
X=[1,2,3]
\]

and multiplying every value by 100:

\[
X'=100X=[100,200,300]
\]

The pattern is unchanged:

\[
1<2<3
\]

and:

\[
100<200<300
\]

Only the scale changed.

A real-world example is changing units:

\[
1\text{ meter}=100\text{ centimeters}
\]

So:

```text
[1, 2, 3] meters
```

becomes:

```text
[100, 200, 300] centimeters
```

The actual physical measurements did not change.

---

## Q9. Why doesn't correlation change when the variable is scaled?

### Answer

Correlation is:

\[
r=\frac{Cov(X,Y)}{\sigma_X\sigma_Y}
\]

Suppose:

\[
X'=100X
\]

Then:

\[
Cov(X',Y)=100Cov(X,Y)
\]

and:

\[
\sigma_{X'}=100\sigma_X
\]

Therefore:

\[
r'=
\frac{100Cov(X,Y)}
{(100\sigma_X)\sigma_Y}
\]

The 100 cancels:

\[
r'=r
\]

So correlation is unaffected by a simple change of scale.

This is one reason correlation is useful for comparing the strength of linear relationships.

---

# Variance and Its Sign

## Q10. Why is variance always positive?

### Answer

Variance is calculated using squared deviations:

\[
Var(X)=\frac{\sum(X_i-\bar X)^2}{n-1}
\]

A square can never be negative:

\[
(-5)^2=25
\]

\[
(+5)^2=25
\]

Therefore every squared deviation is:

\[
\geq0
\]

and the average is also:

\[
\boxed{Var(X)\geq0}
\]

Variance can be **zero**.

For example:

\[
X=[5,5,5,5]
\]

Every value equals the mean, so every deviation is zero:

\[
Var(X)=0
\]

Therefore:

> Variance is never negative. It is either zero or positive.

---

## Q11. Why can covariance be negative while variance cannot?

### Answer

Variance squares one deviation:

\[
(X_i-\bar X)^2
\]

Therefore it cannot be negative.

Covariance multiplies two different deviations:

\[
(X_i-\bar X)(Y_i-\bar Y)
\]

If both deviations have the same sign:

\[
(+)(+)=+
\]

or:

\[
(-)(-)=+
\]

the covariance contribution is positive.

If they have opposite signs:

\[
(+)(-) = -
\]

the covariance contribution is negative.

Therefore covariance can be:

- Positive
- Negative
- Approximately zero

---

# Standard Deviation

## Q12. What is the difference between variance and standard deviation?

### Answer

Both measure the **spread of a variable**, but standard deviation is the square root of variance.

\[
\boxed{SD=\sqrt{Variance}}
\]

For example:

\[
Var(X)=4
\]

Then:

\[
SD(X)=\sqrt4=2
\]

The major difference is their units.

If the original variable is measured in kilograms:

\[
SD=2\text{ kg}
\]

but:

\[
Variance=4\text{ kg}^2
\]

So standard deviation is usually easier to interpret because it is expressed in the **same units as the original variable**.

| | Variance | Standard Deviation |
|---|---|---|
| Measures | Spread | Spread |
| Calculation | Squared deviations | Square root of variance |
| Units | Squared units | Original units |
| Example | \(4\,kg^2\) | \(2\,kg\) |
| Can be negative? | No | No |

---

# Covariance Matrix

## Q13. What is a covariance matrix?

### Answer

A **covariance matrix** is a matrix containing:

1. The variance of every feature on the diagonal.
2. The covariance between every pair of features off the diagonal.

For two features \(X_1\) and \(X_2\):

\[
\boxed{
\begin{bmatrix}
Var(X_1) & Cov(X_1,X_2)\\
Cov(X_2,X_1) & Var(X_2)
\end{bmatrix}
}
\]

For example:

\[
\begin{bmatrix}
4 & 6\\
6 & 9
\end{bmatrix}
\]

means:

```text
             X₁       X₂

X₁            4        6
X₂            6        9
```

Therefore:

- \(Var(X_1)=4\)
- \(Var(X_2)=9\)
- \(Cov(X_1,X_2)=6\)

---

## Q14. Why are variances on the diagonal of a covariance matrix?

### Answer

Because:

\[
\boxed{Var(X)=Cov(X,X)}
\]

So the diagonal represents:

\[
X_1\text{ with }X_1
\]

\[
X_2\text{ with }X_2
\]

and so on.

For example:

\[
\begin{bmatrix}
Var(X_1) & Cov(X_1,X_2)\\
Cov(X_2,X_1) & Var(X_2)
\end{bmatrix}
\]

The diagonal contains variances because each feature is being compared with itself.

---

## Q15. Why are the off-diagonal values covariance?

### Answer

The off-diagonal entries compare **different features**.

For example:

\[
Cov(X_1,X_2)
\]

means:

> How do \(X_1\) and \(X_2\) vary together?

Similarly:

\[
Cov(X_2,X_3)
\]

means:

> How do \(X_2\) and \(X_3\) vary together?

Therefore:

```text
Diagonal
    ↓
Variance of individual features

Off-diagonal
    ↓
Covariance between pairs of features
```

The covariance matrix is symmetric because:

\[
Cov(X,Y)=Cov(Y,X)
\]

---

# Covariance Matrix with 3 Features

## Q16. What does a covariance matrix look like with 3 features?

### Answer

Suppose:

- \(X_1\) = Age
- \(X_2\) = Salary
- \(X_3\) = Experience

The covariance matrix is:

\[
\begin{bmatrix}
Var(X_1) & Cov(X_1,X_2) & Cov(X_1,X_3)\\
Cov(X_2,X_1) & Var(X_2) & Cov(X_2,X_3)\\
Cov(X_3,X_1) & Cov(X_3,X_2) & Var(X_3)
\end{bmatrix}
\]

For example:

\[
\begin{bmatrix}
25 & 100 & 15\\
100 & 40000 & 500\\
15 & 500 & 9
\end{bmatrix}
\]

The diagonal:

```text
25     → variance of Age
40000  → variance of Salary
9      → variance of Experience
```

The off-diagonal:

```text
100 → covariance between Age and Salary
15  → covariance between Age and Experience
500 → covariance between Salary and Experience
```

---

# Why Covariance Matrix Matters in PCA

## Q17. Why do we need a covariance matrix in machine learning?

### Answer

With many features, there are many pairs of variables.

For example, with:

```text
Age
Salary
Experience
Education
```

we may want to know:

- How much Age varies
- How much Salary varies
- How Age and Salary vary together
- How Age and Experience vary together
- How Salary and Experience vary together
- etc.

A covariance matrix stores all of this information in one structure.

For \(n\) features, the covariance matrix has dimensions:

\[
n\times n
\]

It is especially important in **PCA**.

The basic PCA process involves:

\[
\boxed{
Dataset
\rightarrow
Covariance\ Matrix
\rightarrow
Eigenvalues/Eigenvectors
\rightarrow
Principal\ Components
}
\]

In PCA:

- **Eigenvectors** determine the principal directions.
- **Eigenvalues** tell us how much variance exists along those directions.

````python
import numpy as np
import pandas as pd

np.random.seed(23) 

mu_vec1 = np.array([0,0,0])
cov_mat1 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class1_sample = np.random.multivariate_normal(mu_vec1, cov_mat1, 20)

df = pd.DataFrame(class1_sample,columns=['feature1','feature2','feature3'])
df['target'] = 1

mu_vec2 = np.array([1,1,1])
cov_mat2 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class2_sample = np.random.multivariate_normal(mu_vec2, cov_mat2, 20)

df1 = pd.DataFrame(class2_sample,columns=['feature1','feature2','feature3'])

df1['target'] = 0

df = df.append(df1,ignore_index=True)

df = df.sample(40)
````

````python
df.head()
````

````python
# Step 1 - Apply standard scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

df.iloc[:,0:3] = scaler.fit_transform(df.iloc[:,0:3])
````

````python
# Step 2 - Find Covariance Matrix
covariance_matrix = np.cov([df.iloc[:,0],df.iloc[:,1],df.iloc[:,2]])
print('Covariance Matrix:\n', covariance_matrix)
````

````python
# Step 3 - Finding EV and EVs
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)
````

````python
%pylab inline

from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d import proj3d
from matplotlib.patches import FancyArrowPatch


class Arrow3D(FancyArrowPatch):
    def __init__(self, xs, ys, zs, *args, **kwargs):
        FancyArrowPatch.__init__(self, (0,0), (0,0), *args, **kwargs)
        self._verts3d = xs, ys, zs

    def draw(self, renderer):
        xs3d, ys3d, zs3d = self._verts3d
        xs, ys, zs = proj3d.proj_transform(xs3d, ys3d, zs3d, renderer.M)
        self.set_positions((xs[0],ys[0]),(xs[1],ys[1]))
        FancyArrowPatch.draw(self, renderer)

fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(111, projection='3d')

ax.plot(df['feature1'], df['feature2'], df['feature3'], 'o', markersize=8, color='blue', alpha=0.2)
ax.plot([df['feature1'].mean()], [df['feature2'].mean()], [df['feature3'].mean()], 'o', markersize=10, color='red', alpha=0.5)
for v in eigen_vectors.T:
    a = Arrow3D([df['feature1'].mean(), v[0]], [df['feature2'].mean(), v[1]], [df['feature3'].mean(), v[2]], mutation_scale=20, lw=3, arrowstyle="-|>", color="r")
    ax.add_artist(a)
ax.set_xlabel('x_values')
ax.set_ylabel('y_values')
ax.set_zlabel('z_values')

plt.title('Eigenvectors')

plt.show()
````
