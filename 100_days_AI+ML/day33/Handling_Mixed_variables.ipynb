```python
# extract numerical part 
df['Cabin_num'] = df['Cabin'].str.extract('(\d+)')
```

In a regular expression, `\d+` means:

- `\d` matches any digit character (`0` through `9`)
- `+` means “one or more” of the preceding item

So `\d+` matches a sequence of one or more digits, such as `1`, `24`, `1234`, etc.

## What these are called

- These are called **regular expressions** or **regex**.
- In Python, they are used with the `re` module and with pandas string methods like `str.extract()`.

## What `\d+` means

- `\d` = any digit character (`0`–`9`)
- `+` = one or more repetitions
- So `\d+` matches a sequence of one or more digits, e.g. `1`, `24`, `123`

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])

df.head(20)
```


1. `# extract the last bit of ticket as number`
   - This is a comment. It describes the goal of the next few lines: take the last part of the ticket string and turn it into a numeric value.

2. `df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])`
   - For each value in the `Ticket` column:
     - `s.split()` splits the string on whitespace into a list of parts.
     - `[-1]` selects the last part of that list.
   - The result is stored in a new column called `ticket_num`.
   - Example: `"A/521171"` becomes `"A/521171".split()[-1]`, which is `"A/521171"` if there is only one part; if the ticket is `"PC 17599"`, it becomes `"17599"`.

3. `df['ticket_num'] = pd.to_numeric(df['ticket_num'],`
4. `                                   errors='coerce',`
5. `                                   downcast='integer')`
   - These three lines convert the new `ticket_num` column from text to numeric values.
   - `pd.to_numeric(...)` tries to convert each value to a number.
   - `errors='coerce'` means that if a value cannot be converted, it becomes `NaN` instead of causing an error.
   - `downcast='integer'` tries to use a smaller integer type if possible to save memory.
   - After this, `ticket_num` is numeric where conversion succeeded, and `NaN` where it did not.

6. `# extract the first part of ticket as category`
   - Another comment. This one explains the next block: capture the first part of the ticket string, which may represent a category or prefix.

7. `df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])`
   - For each ticket string:
     - `s.split()[0]` splits on whitespace and takes the first part.
   - That first part is stored in a new column called `ticket_cat`.
   - Example: `"PC 17599"` becomes `"PC"`, and `"STON/O2. 3101282"` becomes `"STON/O2."`.

8. `df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,`
9. `                              df['ticket_cat'])`
   - This line cleans the `ticket_cat` values.
   - `df['ticket_cat'].str.isdigit()` checks whether the extracted category is made only of digits.
   - `np.where(..., np.nan, df['ticket_cat'])` means:
     - if the first part is purely numeric, replace it with `NaN`
     - otherwise keep the original category string
   - This is useful because if a ticket started with a number, that first part is not really a category prefix.

10. `df.head(20)`
    - This displays the first 20 rows of the DataFrame so students can see the new `ticket_num` and `ticket_cat` columns and verify the result.

Summary:
- `ticket_num` attempts to capture numeric ticket identifiers.
- `ticket_cat` attempts to capture prefix/category labels from ticket strings.
- Non-numeric prefixes are kept, while numeric-only prefixes are turned into `NaN`.
