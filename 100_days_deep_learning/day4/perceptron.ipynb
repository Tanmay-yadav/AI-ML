# What is perceptron?
- Perceptron is an algorithm which is used for supervised machine Learning.
- we can also call it a mathematical model or mathematical function.
  
- For doing prediction with perceptron we have to know the value of bias and weights.

    ## Neuron vs Perceptron
  - Neuron is a building block of human nervous system and perceptron is the building block of Neural Networks.
  - Perceptron is weekly inspired by  neuron of nervous system, but not exactly same.
    
    
    ## Interpretation
    From Perceptrons we can interpret that weights tell the importance of the feature.

    ## Geometric Intuition 
    - Perceptron works as a separator line between the two outcomes.
    - perceptron can only be used on linear datasets and it is true in 2d,d,4d and so on.
    - perceptron fails in non linear datasets.
    
    


# How to train a perceptron?

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('placement.csv')
```

```python
print(df.shape)
df.head()
```

```python
sns.scatterplot(x=df['cgpa'], y=df['iq'], hue=df['placement'])
```

```python
X =df.iloc[:,0:2]
y =df.iloc[:,-1]
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
p.fit(X,y)
```

```python
p.coef_
```

```python
p.intercept_
# In a perceptron, the intercept is basically the bias. 
# It allows the decision boundary to move away from the origin.
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X.values, y.values, clf=p, legend=1) # here legend is only for the class labels, 
# not for the features and for the location of the label info
# X is a dataframe(contains column names with values) which is why we are using .values to convert it into a numpy array. 
```
