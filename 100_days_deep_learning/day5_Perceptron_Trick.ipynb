# Perceptron Trick
`Perceptron` is a model or algorithm which define the line between the two outcomes. 

- In perceptron we take a random datapoint and check if it is correctly classified or not.
- When the datapoint is misclassified then we move the line towards that point. we repeat this process until convergence.
  

- In perceptron Trick we add 1 at last to the random point (like point(4,3)-->becomes(4,3,1) this one is kind of a bias value )that we pick and subtract(if the datapoint is misclassifed as True) or add(when datapoint is misclassified as False).
