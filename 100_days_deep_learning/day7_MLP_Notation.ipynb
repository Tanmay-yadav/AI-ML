# MLP Notation 


- There are 4 layers and the data we have here is 4 dimensional.
- `Data -> {m,n}`
  - m -> rows
  - n =column i.e 4
  - each row denoted by $$x_{i}$$
- `Trainable parameters`: The first thing that we do is to find the trainable parameters i.e total number of weights and bias.
  - The weights between input layer and hidden layer are 4*3=12 weights and 3 bias, and similarly in other layers also as shown above.
- `Notations`
  - `Bias`: $$b_{ij}$$ --> here i represents layer number and j represent the node number.
  - `Output`: $$O_{ij}$$ --> i is layer number and j is node number.
  - `Weights`: $$w_{ij}^{(k)}$$
    - `K`: konse layer mein ja rha hai
    - `i`:pichle waale layer mein konse node se nikal rha hai
    - `j`:Agle layer mein konse node mein ja rha hai.
