# Accessing array rows and columns

<p>One commonly needed routine is accessing single
rows or columns of an array. You can do this by combining indexing and slicing,
using an empty slice marked by a single colon (:):
</p>

```python
print(x2)
```

```python
print(x2[:,0]) # first column of x2
```

```python
print(x2[0,:]) # first row of x2
```

```python
# in case of row access, the empty slice can be omitted for a more compact syntax:
print(x2[0]) # equuivalrnt to x2[0,:]
```

### Subarrays as no-copy views

<p>One important—and extremely useful—thing to know about array slices is that they
return views rather than copies of the array data. This is one area in which NumPy
array slicing differs from Python list slicing: in lists, slices will be copies. Consider our
two-dimensional array from before:</p>

```python
print(x2)
```

```python
# let's extract a 2X2 sub-array from x2
x2_sub = x2[:2,:2]
print(x2_sub)
```

```python
# now if we modify this subarray, we'll see that the original array is changed! observe:
x2_sub[0,0] = 99
print(x2_sub)
```

```python
print(x2)
# This default behavior is actually quite useful: it means that when we work with large
# datasets, we can access and process pieces of these datasets without the need to copy
# the underlying data buffer.
```

## creating copies of arrays 
<p>Despite the nice features of array views, it is sometimes useful to instead explicitly
copy the data within an array or a subarray. This can be most easily done with the
copy() method:
</p>

```python
x2_sub_copy = x2[:2,:2].copy()
print(x2_sub_copy)
```

```python
x2_sub_copy[0,0]= 42
print(x2_sub_copy)
```

```python
print(x2) 
# here we see that our copy operation worked as expected, and the original array is unchanged
```

# Array Concatenation and Splitting 
<p>All of the preceding routines worked on single arrays. It’s also possible to combine
multiple arrays into one, and to conversely split a single array into multiple arrays.
We’ll take a look at those operations here.</p>

## Concatenation of arrays 
<p>Concatenation, or joining of two arrays in NumPy, is primarily accomplished
through the routines np.concatenate, np.vstack, and np.hstack. np.concatenate
takes a tuple or list of arrays as its first argument, as we can see here:
</p>


```python
x= np.array([1,2,3])
y= np.array([3,2,1])
np.concatenate((x,y))
```

```python
z=[99,99,99]
print(np.concatenate((x,y,z)))
```

```python
# np.concatenate can also be used for two-dimensional arrays:
grid= np.array([[1,2,3],
                [4,5,6]])
# concatenate along the first axis (rows)
np.concatenate((grid,grid))
```

```python
# concatenate along the second axis (columns)
np.concatenate((grid,grid), axis=1)
```

```python
# for Working of mixed dimesions, it can be clearer to use the np.vstack and np.hstack functions:
# vertical stack
x= np.array([1,2,3])
grid = np.array([[9, 8, 7],
 [6, 5, 4]])
print("Vertical stack result:",np.vstack((x,grid)))
# horizontal stack
y= np.array([[99],
 [99]])
np.hstack((grid,y))
print("Horizontal stack result:",np.hstack((grid,y)))
```

## Splitting of arrays 
<p>The opposite of concatenation is splitting, which is implemented by the functions
np.split, np.hsplit, and np.vsplit. For each of these, we can pass a list of indices
giving the split points:
</p>

```python
x=[1,2,3,99,99,3,2,1]
x1,x2,x3=np.split(x,[3,5])
print(x1,x2,x3)
```

<p>Notice that N split points lead to N + 1 subarrays. The related functions np.hsplit
and np.vsplit are similar:</p>

```python
grid= np.arange(16).reshape((4,4))
print("Original grid:\n",grid)
```

```python
upper, lower= np.vsplit(grid,[2])
print("Upper part:\n",upper)    
print("Lower part:\n",lower)
```

```python
left, right= np.hsplit(grid,[2])
print("Left part:\n",left)
print("Right part:\n",right)
```

# Computation on NumPy Arrays:Universal Functions 
<p>Computation on NumPy arrays can be very fast, or it can be very slow. The key to
making it fast is to use vectorized operations, generally implemented through Num‐
Py’s universal functions (ufuncs). This section motivates the need for NumPy’s ufuncs,
which can be used to make repeated calculations on array elements much more effi‐
cient. It then introduces many of the most common and useful arithmetic ufuncs
available in the NumPy package.</p>

## The Slowness of loops 
<p>The relative sluggishness of Python generally manifests itself in situations where
many small operations are being repeated—for instance, looping over arrays to operate on each element.</p>

```python
import numpy as np
np.random.seed(0) # what is seed?--> seed is used to initialize the random number generator, ensuring reproducibility
def compute_reciprocals(values):
    output =np.empty(len(values))
    for i in range(len(values)):
        output[i]=1.0/values[i]
    return output 
values=np.random.randint(1,10,size=5) # her 1 is inclusive and 10 is exclusive 
compute_reciprocals(values) 
```

```python
big_array = np.random.randint(1,100,size=1000000)
%timeit compute_reciprocals(big_array)
compute_reciprocals(big_array)
```

# introduccing UFuncs
<p>For many types of operations, NumPy provides a convenient interface into just this
kind of statically typed, compiled routine. This is known as a vectorized operation.
You can accomplish this by simply performing an operation on the array, which will
then be applied to each element. This vectorized approach is designed to push the
loop into the compiled layer that underlies NumPy, leading to much faster execution.</p>

```python
print(compute_reciprocals(values))
print(1.0/values)
```

```python
%timeit (1.0/big_array)
```

```python
# Vectorized operations in NumPy are implemented via ufuncs, whose main purpose is
# to quickly execute repeated operations on values in NumPy arrays. Ufuncs are
# extremely flexible—before we saw an operation between a scalar and an array, but we
# can also operate between two arrays:
np.arange(5)/np.arange(1,6)
```

```python
x=np.arange(9).reshape((3,3))
2 **x # exponentiation means raise to the power, here 2 raised to the power of each element in x
```

```python
# ufunc operations are not limited to 
# one-dimensional arrays--they can act on 
# multidimensional arrays as well:
x= np.arange(9).reshape((3,3))
2 **x 
```
