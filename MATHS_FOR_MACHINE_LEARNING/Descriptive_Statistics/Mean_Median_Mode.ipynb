# Measures of central Tendency

```python
import random
import time
import numpy as np
from collections import Counter
from scipy import stats # for mode
```

```python
# Generte 10,000 random integers between 1 and 1000
data =[random.randint(1,1000)for _ in range(1000000)]
np_data=np.array(data)
```

```python
#--------------------------------------------
# 1. From Scratch Implementation 
#--------------------------------------------
def mean_from_scratch(arr):
    return sum(arr)/len(arr)

def median_from_scratch(arr):
    sorted_arr =sorted(arr)
    n=len(sorted_arr)
    mid=n//2
    if n%2==0:
        return (sorted_arr[mid-1]+sorted_arr[mid])/2
    else:
        return sorted_arr[mid]
    
def mode_from_scratch(arr):
    freq=Counter(arr)
    max_freq =max(freq.values())
    modes=[k for k,v in freq.items() if v==max_freq]
    return modes[0] if len(modes)== 1 else modes 
```

```python
#Timing the from scratch implementation
start_time= time.time()
mean_s=mean_from_scratch(data)
median_s =median_from_scratch(data)
mode_s=mode_from_scratch(data)
end_time=time.time()
Scratch_time=end_time-start_time
```

```python
#--------------------------------------------
# 2. Using Numpy Built-in Implementation
#--------------------------------------------
start =time.time()
mean_np=np.mean(np_data)
median_np=np.median(np_data)
mode_np=stats.mode(np_data,keepdims=True).mode[0] # scipy mode as numpy does not have a built-in mode function
end=time.time()
numpy_time=end-start
```

```python
#--------------------------------------------
# Results
#--------------------------------------------
print("From Scratch:")
print(f"Mean:{mean_s}")
print(f"Median:{median_s}")
print(f"Mode:{mode_s}")
print(f"Time:{Scratch_time}")

print("\nUsing NumPy:")
print(f"Mean:{mean_np}")
print(f"Median:{median_np}")
print(f"Mode:{mode_np}")
print(f"Time:{numpy_time}")
print("Speedup:", Scratch_time/numpy_time)
```
