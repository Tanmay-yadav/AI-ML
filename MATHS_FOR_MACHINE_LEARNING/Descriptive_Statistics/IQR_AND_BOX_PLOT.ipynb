```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# generate 10,00,000 random numbers 
np.random.seed(42) 
data =np.random.randint(1,1001,size=1000000) # random integers 1-1000
```

- **What it does:** `random.seed(x)` initializes the PRNG state so outputs are deterministic.
- **Argument:** pass an integer, string, or `None` (`None` uses system entropy).
- **Reproducibility:** same seed → same sequence (useful for tests, experiments).
- **When to use:** debugging, demos, ML experiments, unit tests, examples.
- **When not to use:** cryptographic needs — use `secrets` or `random.SystemRandom`.
- **Example:** `random.seed(42)` → repeatable `random.random()` outputs across runs.

# Percentile
 * percentile isa value below which a given percentage of data fall.
 * ranges from [0,100]
 * if p(25)=x , it means 25% of our sample lies below value x

```python
#--------------------------------------------------
#  1. Percentile Built-in Implementation
#--------------------------------------------------
p25 =np.percentile(data,25) # 25th percentile(Q1)
p50 =np.percentile(data,50) # 50th percentile(Q2)
p75 =np.percentile(data,75) # 75th percentile(Q3)
print(f"25th Percentile(Q1): {p25}")
print(f"50th Percentile(Q2): {p50}")    
print(f"75th Percentile(Q3): {p75}")
```



- **Definition:** a formatted string literal (f-string) is a string prefixed with `f` or `F` that allows expressions inside `{}` to be evaluated and inserted.  
- **Syntax:** `f"text {expression} more"` — the `{expression}` is replaced by its value at runtime.  
- **Introduced:** Python 3.6.  
- **Expressions allowed:** variables, arithmetic, function calls, attribute access, e.g. `f"{x:.2f}", f'{user.name.upper()}'`.  
- **Formatting:** use `:` inside braces for format specs, e.g. `f"{value:,.2f}"`.  
- **Literal braces:** use `{{` or `}}` to include `{` or `}` literally.  
- **When to use:** concise, readable string interpolation for printing, logging, and formatted output.  
- **Caveat:** expressions are evaluated at runtime — avoid embedding untrusted code or complex side-effecting expressions.


- **Snippet:** `print(f"25th Percentile(Q1): {p25}")`  
- **`f` prefix:** makes the string a formatted string literal (f-string).  
- **`{}` meaning:** any valid Python expression inside `{}` is evaluated and its value inserted.  
- **Formatting inside `{}`:** use `:` to format, e.g. `f"{p25:.2f}"` for two decimals.  
- **Expressions allowed:** you can put expressions, e.g. `f"{p25/2}"` or `f"{min(a,b):.1f}"`.  
- **Literal braces:** write `{{` or `}}` to include a literal `{` or `}`.  
- **When evaluated:** evaluation happens at runtime when the f-string is created.  
- **Why use f-strings:** concise, readable, and generally faster than `%` or `str.format()`; available from Python 3.6+.

# Quantile 
* quantile is a value below which a given fraction (0-1) of data falls.
* ranges from [0,1]
* if Q(0.25)= x , then it means 1/4th of our sample lies below x

```python
# ----------------------------------------
# Quantile Built-in Implementation
# ----------------------------------------
q1 =np.quantile(data,0.25) # 25th percentile(Q1)
q2 =np.quantile(data,0.50) # 50th percentile(Q2)
q3 =np.quantile(data,0.75) # 75th percentile(Q3)
print(f"25th Quantile(Q1): {q1}")
print(f"50th Quantile(Q2): {q2}")
print(f"75th Quantile(Q3): {q3}")
```

# Quartile
* Quartiles divide order data into four equla parts, giving three key cut points:
* Q1(1st Quartile)= 25th percentile (lower quartile)=25% of the data lie below Q1
* Q2(2nd Quartile)= 75th percentile (upper Quartile)=75% lie below Q3

# InterQuartile Range(IQR) 
* It REpresents the middle 50% data.
* IQR =Q3-Q1

```python
#--------------------------------------------
# IQR Built-in Implementation
#--------------------------------------------
Q1 =np.percentile(data,25) # 25th percentile(Q1)
Q3 =np.percentile(data,75) # 75th percentile(Q3)
IQR = Q3 - Q1
print(f"Q1(25th Percentile): {Q1}")
print(f"Q3(75th Percentile): {Q3}")
print(f"Interquartile Range(IQR): {IQR}")
```

# Box Plot 

* Any data point below min and above max can be considered an oultier 

```python
# ---------------------------
# BOX-plot Implementation
# ---------------------------
plt.figure(figsize=(10,5))
sns.boxplot(data=data, orient='h', color='lightblue',linewidth=1)
plt.title('Box Plot using seaborn ')
plt.ylabel('Values')
plt.show() 
```
