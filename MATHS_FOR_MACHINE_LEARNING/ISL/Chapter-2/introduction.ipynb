<p>We now create some more sophisticated plots. The ax.contour() method .contour() produces a contour plot in order to represent three-dimensional data, similar contour plot to a topographical map. It takes three arguments:</p>
<p>• A vector of x values (the first dimension),
   • A vector of y values (the second dimension), and
   • A matrix whose elements correspond to the z value (the third dimension) for each pair of (x,y) coordinates.</p>
<p>To create x and y, we’ll use the command np.linspace(a, b, n), which np.linspace() returns a vector of n numbers starting at a and ending at b.</p>

```python
fig, ax = subplots(figsize=(8, 8))
x = np.linspace(-np.pi, np.pi, 50)
y=x
f = np.multiply.outer(np.cos(y), 1 / (1 + x**2))
ax.contour(x, y, f);
```
